# Customer RFM Segmentation & Statistical Hypothesis Testing
**Objective:** Perform an independent two-sample t-test to determine if the spend difference between 'Champions' and 'At Risk' segments is statistically significant.


```python
import pandas as pd
```

```python
#loading the file
df = pd.read_csv('customers_segments.csv')
```

```python
#Top 10 rows
df.head(10)
```

```python
# count customer_segment  
print(df['customer_segment'].value_counts())
```

```python
# Separate the "monetary" values for each segment we want to compare
champions = df[df['customer_segment'] == 'Champions']['monetary']
at_risk = df[df['customer_segment'] == 'At Risk']['monetary']

#Quick check
print(champions.mean())
print(at_risk.mean())
```

```python
# Import the statistics module for hypothesis testing
from scipy import stats

# Welch's t-test: equal_var=False since the two groups likely have
# different variances (different sizes and spread)

t_stat, p_value = stats.ttest_ind(champions, at_risk, equal_var = False)

print(f"t-statistic: {t_stat}")
print(f"p-value: {p_value}")
```

```python
# Final statistical interpretation
alpha = 0.05

if p_value < alpha:
    print(f"Result: Statistically Significant (p-value = {p_value:.4e} < {alpha})")
    print("Conclusion: Champions spend significantly more than At Risk customers.")
else:
    print(f"Result: Not Statistically Significant (p-value = {p_value:.4f} >= {alpha})")
    print("Conclusion: No significant difference in spend detected between segments.")
```
